# D3 Radial Equilibrium Equation

## Introduction

The pre-design studied in notebooks D1 and D2 give a first approach of the fan performance, focused on a 1D analysis. But they are not suitable for the complex flow produced
along the blade span in the axial fan. Specifically, it is expected to find a very different behavior near the hub and the tip of the fan, since the
actual flow in the turbomachine is three-dimensional.

The rotation of the flow downstream produces a **pressure gradient** along the radial direction, in order to balance the centrifugal force.
That leads to a redistribution of the axial 
velocity, $c_x$,
and, hence, of the flow rate. Besides, the pressure is also not uniform, and the performance of the fan has to be estimated by integrating both
the axial velocity and pressure. These two variables distributions are governed by the **Radial Equilibrium Equation (REE)**, that relates 
the derivatives on the radial equation with the profile of the _tangential velocity_, $c_\theta$, that is defined by the blade twisting and,
therefore, by the design of the fan.


## Learning Objectives

- **Understand** the importance of computing 3D distribution of velocity and pressure
- **Interpret** the Radial Equilibrium Equation (REE)
- **Solve the Inverse (Design) problem** by using the REE to determine the 3D blade design for a desired swirl distribution of flow.
- **Solve the Direct (Analysis) problem** by using the REE to infer fan performance from a blade design.
- **Identify** the physical onset of aerodynamics stall and flow reverse on off-design operating conditions.

### 1- Physical interpretation of REE
Write down the basic radial force balance equation for a fluid element in a swirling flow (Equation 5.11). Explain briefly, in your own words, what physical forces are in balance here and why a radial pressure gradient ($dp/dr$) must exist downstream of a fan rotor.
 
Answer: 

**Equation 5.11:**
$$\frac{1}{\rho}\frac{dp}{dr} = \frac{c_\theta^2}{r}$$

The forces in equilibrium here are the radial pressure gradient $\frac{dp}{dr}$ and the angular momentum of the fluid $\frac{c_\theta^2}{r}$, density is assumed as constant due an incompressible hypotesis. We observe the balance between the centripetal acceleration force, to maintain it's curve trajectory. This force will push fluid to outwards, to prevent the fluid from slosh to the casing and to maintain steady axisymetric without unbalances, we need a force to counteract it. In this case, the increase in pressure towards outer radius.

### 2- Design versus Analysis approaches
The two approaches for using the radial equilibrium equation are the **Inverse** and the **Direct** one. Describe the theoretical difference between them:
- What information is the "input" for the **Design (Inverse)** problem, and what is the output?
- In the **Analysis (Direct)** problem, what geometric parameter of the blade is considered fixed, and what resulting flow field characteristics are we trying to predict?

Answer:

**Design (Inverse) problem:** As Input we observe the desired flow specification and thermodynamic distributions, specifically target swirl velocity $c_\theta(r)$ (or stagnation enthalpy $h_o(r)$) and entropy/losses $s(r)$. As output, the required axial velocity profile $c_{x}(r)$, velocity triangles, flow angles across the span, and the resulting blade geometry/profiles needed to achieve them.

**Analysis (Direct) Problem:** As Input we observe as fixed geometric parameter of the blade the existing blade profile geometry (specifically the relative or absolute flow outlet angle, ($\alpha_2(r)$ or $\beta_2(r)$). Characteristics to predict are the resulting velocity field, specifically the axial velocity profile $c_x(r)$, downstream swirl distribution $c_\theta(r)$, and overall fluid dynamic performance.


### 3- The free-vortex special case
In a **Free-Vortex** swirl distribution ($c_\theta \cdot r = \text{constant}$):
- What happens to the axial velocity profile $c_x(r)$ downstream? Explain using the Radial Equilibrium Equation (Equation 5.14) or Lewis's derivation.
- What specific aerodynamic or geometric difficulty does this law typically create at the **blade hub** for a single-stage fan?

Answer:

**To the Axial Velocity Profile $c_x(r)$ downstream**, we observe for free-vortex flow ($c_\theta \cdot r = \text{constant}$), the axial velocity profile downstream remains uniform across the radius ($c_x(r) = C_x = \text{constant}$). Under radial equilibrium:
   $$\frac{1}{\rho}\frac{dp_o}{dr} = c_x \frac{dc_x}{dr} + \frac{c_\theta}{r}\frac{d(r c_\theta)}{dr}$$

Since constant work input gives $dp_o/dr = 0$ and free-vortex swirl gives $d(r c_\theta)/dr = 0$, it simplifies directly to $c_x (dc_x/dr) = 0$, meaning $c_x$ is independent of radius. 

To the Aerodynamics or Geometry diffuculty, at small radius (near the hub), $r$ is small, forcing the required tangential velocity $c_\theta = K/r$ and stage loading/pressure rise coefficient $\psi \propto 1/r^2$ to become excessively large. This leads to two main problems, the first one is extreme fluid deflection angles ($\epsilon_R$), leading to severe aerodynamic stall/diffusion risks, the other one is the small pitch-to-chord ratios ($t/l$), requiring huge blade chords at the root that are physically difficult to fit onto the rotor hub.  


### 4- The forced-vortex special case
For a **Forced-Vortex / Solid-Body** swirl distribution ($c_\theta \propto r$):
- Does the axial velocity $c_x$ increase or decrease as you move from the hub to the tip? Refer to Lewis Section 5.3.1 or Example 5.2 to support your answer.
- Why might a designer choose this distribution over a free-vortex despite it producing non-uniform axial flow?

Answer:

For a forced-vortex distribution ($c_\theta \propto r$), the axial velocity $c_x$ decreases as you move from the hub to the tip. As shown in Example 5.2 and Table 5.3, $c_x$ is maximum at the hub ($r_h$) and drops to a minimum at the tip ($r_t$).

![Figure 5.3: Table that shows that cx values is inversal to raidus value ](images/TABLE_5_3_CX.png)

Designers choose forced-vortex distributions because they unload the blade hub region. By reducing the work input demand at the root and shifting loading toward the larger radius, it avoids the extreme blade twist, massive root chords, and stall-prone high deflections associated with free-vortex hubs, allowing lower hub-to-tip ratios and larger flow areas.

### 5- Advantages of mixed-vortex law
Theoretical designs often use a **Mixed-Vortex law** ($c_\theta = \frac{a}{r} + br$).
- How does this law allow a designer to combine the benefits of both free-vortex and forced-vortex flows? [8, 9]
- How can the parameters $a$ and $b$ be used to control the work input (pressure rise) distribution across the blade span?

Answer:

The mixed-vortex law ($c_\theta = \frac{a}{r} + br$) combines free-vortex ($a/r$) and forced-vortex ($br$) components. It maintains reasonable root aerodynamic loading and pitch/chord ratios near the hub (forced-vortex benefit) while keeping meridional axial velocity variations and tip loading manageable across the span (free-vortex benefit).

The specific work input coefficient is given by $\psi(r) = \frac{1}{\Omega}\left(\frac{a}{r^2} + b\right)$. So, parameter $a$ sets the free-vortex weight, controlling the steepness of the $1/r^2$ radial loading gradient. For the $b$ parameter, adds a uniform baseline work addition across all radius.

By adjusting $a$ and $b$ (or specifying values $\psi_m$ at mean radius and $\psi_0$ at a reference radius), a designer can precisely tailor how work/pressure rise is split between the hub and tip to prevent hub stalling while maximizing overall stage power capacity.
